## Question 6

[Task 2.2] In `plot_histogram`, what does `sort='value_desc'` do?

**A.** Sorts bars by key descending  
**B.** Sorts bars by count descending  
**C.** Sorts by Hamming weight  
**D.** Sorts by shot index  

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: B**

**Explanation:**  
`value_desc` orders outcomes by frequency from highest to lowest.
Note that `desc` would orders outcomes by keys (label) from highest to lowest.

</details>
