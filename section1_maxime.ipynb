## Question 1

For a 2-qubit system in Qiskit, which Pauli label applies Z on qubit 1 and X on qubit 0?

**A.** `'XZ'`  
**B.** `'ZX'`  
**C.** `'ZI'`  
**D.** `'IX'`

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: B**

**Explanation:**  
Qiskit Pauli strings are right-to-left by qubit index (rightmost char is qubit 0).  
So `Z` on qubit 1 and `X` on qubit 0 is `'ZX'`.  
- `'XZ'` would mean X on qubit 1 and Z on qubit 0.

</details>

## Question 2

Which statement is correct for the commutator `[X, Z] = XZ - ZX`?

**A.** `[X, Z] = 0`  
**B.** `[X, Z] = 2iY`  
**C.** `[X, Z] = -2iY`  
**D.** `[X, Z] = I`

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: C**

**Explanation:**  
Using Pauli algebra: `XZ = -iY` and `ZX = iY`, so `XZ - ZX = -2iY`.  
- **A** is false (they anti-commute).  
- **B** has the wrong sign.  
- **D** is unrelated.

</details>

## Question 4

For the Bell state `|Phi+> = (|00> + |11>)/sqrt(2)`, what is `<XX>`?

**A.** `-1`  
**B.** `0`  
**C.** `1`  
**D.** `0.5`

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: C**

**Explanation:**  
`|Phi+>` is a +1 eigenstate of both `ZZ` and `XX`, so `<XX> = 1`.  
This is a standard two-qubit correlation property used in many exam questions.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, SparsePauliOp
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
sv = Statevector(qc)
print(sv.expectation_value(SparsePauliOp('XX')).real)  # 1.0
```

</details>

## Question 7

Given `P = Pauli('Y')`, which statement is true?

**A.** `P` is unitary but not Hermitian  
**B.** `P` is Hermitian but not unitary  
**C.** `P` is both unitary and Hermitian  
**D.** `P` is neither unitary nor Hermitian

<details>
<summary>Click to reveal Answer & Explanation</summary>

## Question 8

Which `SparsePauliOp` expression is dimensionally valid for a 3-qubit Hamiltonian?

**A.** `SparsePauliOp.from_list([('ZZ', 0.7), ('XII', 0.2)])`  
**B.** `SparsePauliOp.from_list([('ZIZ', 0.7), ('XII', 0.2)])`  
**C.** `SparsePauliOp.from_list([('Z', 0.7), ('XII', 0.2)])`  
**D.** `SparsePauliOp.from_list([('IIII', 1.0), ('XII', 0.2)])`

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: B**

**Explanation:**  
All Pauli strings in one `SparsePauliOp` must have consistent length equal to qubit count.  
- **B** uses only 3-character strings.  
- **A** and **C** mix 2/1 with 3.  
- **D** mixes 4 with 3.

</details>

## Question 9

Consider:

```python
qc = QuantumCircuit(1)
qc.y(0)
qc.y(0)
print(Statevector(qc))
```

What is printed?

**A.** `Statevector([1.+0.j, 0.+0.j])`  
**B.** `Statevector([-1.+0.j, 0.+0.j])`  
**C.** `Statevector([0.+0.j, 1.+0.j])`  
**D.** `Statevector([0.+1.j, 0.+0.j])`

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: A**

**Explanation:**  
Pauli-Y is self-inverse: `Y^2 = I`.  
Starting from `|0>`, two Y gates return `|0>`, giving `[1, 0]`.

</details>

## Question 10

Which statement about global phase is correct in Qiskit operator comparisons?

**A.** Two states/operators differing only by global phase represent different physical states  
**B.** Global phase can change measurement probabilities for a closed system  
**C.** `Operator.equiv` treats globally phased operators as equivalent  
**D.** Global phase only appears for multi-qubit states

<details>
<summary>Click to reveal Answer & Explanation</summary>

**Correct Answer: C**

**Explanation:**  
Global phase does not affect observable probabilities for an isolated state.  
Qiskit's `.equiv()` is specifically designed for equivalence up to global phase.

```python
from qiskit.quantum_info import Operator
I = Operator.from_label('I')
minus_I = -1 * I
print(I.equiv(minus_I))  # True
```

</details>
